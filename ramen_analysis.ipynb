```python
import pandas as pd
import numpy as np

# ==========================================
# 1. LOAD YOUR DATA
# ==========================================
print("Loading Ramen dataset...")
df = pd.read_csv('ramen_ratings.csv')

# ==========================================
# 2. GET TO KNOW ITS STRUCTURE
# ==========================================
print("\n--- First 5 rows ---")
display(df.head()) 

print("\n--- Dataset Dimensions ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

print("\n--- Dataset Info (Types & Nulls) ---")
df.info()

# ==========================================
# 3. EXPLORE INDIVIDUAL COLUMNS
# ==========================================
print("\n--- Most Common Packaging Styles ---")
display(df['Style'].value_counts())

print("\n--- Top 10 Countries by Number of Reviews ---")
display(df['Country'].value_counts().head(10))

# ==========================================
# 4. BASIC CLEANING AND WRANGLING
# ==========================================
print("\n--- Cleaning & Wrangling Process ---")

print("Missing values per column before cleaning:")
print(df.isna().sum())

duplicate_count = df.duplicated().sum()
print(f"\nDropping {duplicate_count} duplicate rows...")
df.drop_duplicates(inplace=True)

df.rename(columns={
    'Review #': 'Review_ID',
    'Top Ten': 'Top_Ten_Year'
}, inplace=True)

df['Stars'] = pd.to_numeric(df['Stars'], errors='coerce')

cup_noodles = df[df['Style'] == 'Cup']
print(f"\nFiltered down to {cup_noodles.shape[0]} Cup-style ramens.")

worst_ramen = df.sort_values(by='Stars', ascending=True)

print("\n--- The 5 Lowest Rated Ramens ---")
display(worst_ramen[['Brand', 'Variety', 'Country', 'Stars']].head())

print("\nFirst Steps Complete!")
```
